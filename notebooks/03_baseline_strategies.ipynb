# 03 — Baseline strategies (Phase 4)

Five rule-based benchmarks, generated **before** any ML model:

1. Buy-and-hold equivalent (always long the FX pair)
2. SMA crossover
3. N-bar momentum
4. Rolling z-score mean reversion
5. Donchian breakout of the *prior* N bars

Naive next-bar returns in the report are **not** a backtest (no spread, no stops filled, no position sizing).

```bash
python -m quant_forex ingest --config configs/default.yaml
python -m quant_forex baselines --config configs/default.yaml
```

````python
from quant_forex.config.loader import load_config
from quant_forex.exploration.io import latest_processed_csv
from quant_forex.strategies.pipeline import run_baselines
from quant_forex.utils.paths import ProjectPaths

paths = ProjectPaths.find()
config = load_config(paths.resolve("configs/default.yaml"))
processed = latest_processed_csv(paths.data_processed)
result = run_baselines(config, paths, input_csv=processed)
for run in result.runs:
    d = run.diagnostics
    print(d.strategy, "entries", d.n_entries, "hit", d.naive_hit_rate)
print(result.report_path)
if result.runs[0].signals:
    print(result.runs[0].signals[0].model_dump())
````
