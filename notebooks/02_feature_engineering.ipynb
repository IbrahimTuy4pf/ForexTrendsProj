# 02 — Feature engineering (Phase 3)

Indicators use only data available at the current bar (rolling windows are right-aligned). Warmup NaNs at the start are expected.

Do **not** treat a high in-sample correlation with the next return as a strategy. Feature selection on the full sample is a form of leakage; chronological splits belong in later phases.

```bash
python -m quant_forex ingest --config configs/default.yaml
python -m quant_forex features --config configs/default.yaml
```

````python
from quant_forex.config.loader import load_config
from quant_forex.exploration.io import latest_processed_csv, load_processed_ohlcv
from quant_forex.features.engine import build_features
from quant_forex.features.importance import next_bar_associations
from quant_forex.features.pipeline import run_feature_pipeline
from quant_forex.utils.paths import ProjectPaths

paths = ProjectPaths.find()
config = load_config(paths.resolve("configs/default.yaml"))
processed = latest_processed_csv(paths.data_processed)
ohlcv = load_processed_ohlcv(processed)
built = build_features(ohlcv, config.features)
print(built.groups)
print(len(built.feature_columns), "features")
built.frame.head()
````

````python
assoc = next_bar_associations(built.frame, built.feature_columns)
assoc.head(10)
````

````python
result = run_feature_pipeline(config, paths, input_csv=processed)
print(result.output_csv)
print(result.report_path)
````
