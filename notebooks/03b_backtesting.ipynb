# Backtesting (Phase 5)

Signals at bar **close** fill at the **next open**, with spread and slippage. Same-bar stop and take-profit → stop first.

This is still OHLC research, not tick-level broker fills, and not a live result.

```bash
python -m quant_forex ingest --config configs/default.yaml
python -m quant_forex backtest --config configs/default.yaml
python -m quant_forex backtest --strategy ma_crossover
```

````python
from quant_forex.config.loader import load_config
from quant_forex.backtesting.metrics import compute_metrics
from quant_forex.backtesting.pipeline import run_backtest_pipeline
from quant_forex.exploration.io import latest_processed_csv
from quant_forex.utils.paths import ProjectPaths

paths = ProjectPaths.find()
config = load_config(paths.resolve("configs/default.yaml"))
processed = latest_processed_csv(paths.data_processed)
result = run_backtest_pipeline(config, paths, input_csv=processed)
for item in result.results:
    metrics = compute_metrics(item)
    print(item.strategy, "trades", metrics["number_of_trades"], "return", metrics["total_return"], "halted", item.halted)
print(result.report_path)
````
