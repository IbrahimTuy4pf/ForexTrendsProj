# 07 — Paper trading (Phase 10)

This is a **simulated** book on local OHLCV. It does not log into a broker. Live trading cannot be enabled.

```bash
python -m quant_forex ingest --config configs/default.yaml
python -m quant_forex paper --config configs/default.yaml --enable-paper --strategy ma_crossover
```


````python
from quant_forex.config.loader import load_config
from quant_forex.execution.broker import LiveTradingNotImplemented, submit_live_order
from quant_forex.execution.pipeline import run_paper
from quant_forex.exploration.io import latest_processed_csv
from quant_forex.utils.paths import ProjectPaths

paths = ProjectPaths.find()
config = load_config(paths.resolve("configs/default.yaml"))
config.execution.paper_trading = True
processed = latest_processed_csv(paths.data_processed)
result = run_paper(config, paths, input_csv=processed, strategy_name="ma_crossover")
print(result.report_path)
print(result.equity, result.units, result.pending)
try:
    submit_live_order()
except LiveTradingNotImplemented as exc:
    print(exc)
````
